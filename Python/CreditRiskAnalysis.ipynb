## Ml Exp

##### Exp1.1

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=24,test_size=0.20,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
```

```python
model.coef_
```

```python
def feature_elaborate(x,model):
    for each in range(len(x.columns)):
        print(f'the one unit of {x.columns[each]} increase has this rate of change in prediction {model.coef_.flatten()[each]}')
```

```python
feature_elaborate(x=x_train,model=model)
```

```python
predict_prob=model.predict_proba(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score

y_pred = model.predict(x_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, model.predict_proba(x_test)[:,1]))
```

```python
def CheckForNewCustomer(new_customer):
    nc=new_customer.copy()
    ss_new_customer=ss.transform(new_customer)
    customer=model.predict_proba(ss_new_customer)[:,1]
    cus=pd.DataFrame(customer,columns=['Default in Percent'])
    cusb=pd.DataFrame(customer.round(),columns=['Default 0/1'])
    return pd.concat([nc,cus,cusb],axis=1)
```
